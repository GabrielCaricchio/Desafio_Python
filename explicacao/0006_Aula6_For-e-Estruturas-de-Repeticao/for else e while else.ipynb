### For Else? While Else?

```python
lista_vendas = [7000, 5000, 8500, 9000, 10000, 150, 25]

for venda in lista_vendas:
    if venda > 8000:
        print("Vai pagar imposto")
        break
```

```python
lista_vendas = [7000, 5000, 8500, 9000, 10000, 150, 25]

pagar_imposto = False
for venda in lista_vendas:
    if venda > 11000:
        print("Vai pagar imposto")
        pagar_imposto = True
        break

if pagar_imposto == False:
    print("Não vai pagar imposto")
```

```python
lista_vendas = [7000, 5000, 8500, 9000, 10000, 150, 25]

for venda in lista_vendas:
    if venda > 11000:
        print("Vai pagar imposto")
        break
else: # Só vai ser executado se o for terminar normalmente, ou seja, sem encontrar um break, quando o for é executado completamnete
    print("Não vai pagar imposto")
```

```python
lista_vendas = [7000, 5000, 8500, 9000, 10000, 150, 25]
i=0

while i < len(lista_vendas):
    venda = lista_vendas[i]
    if venda > 11000:
        print("Vai pagar imposto")
        break
    i += 1
else: # Só vai ser executado se o while terminar normalmente, ou seja, sem encontrar um break, quando o while é executado completamnete
    print("Não vai pagar imposto")
```
