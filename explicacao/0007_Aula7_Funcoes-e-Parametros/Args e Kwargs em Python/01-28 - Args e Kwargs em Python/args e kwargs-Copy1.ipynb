```python
def calcular_imposto(valor, percentual_imposto_renda):
    ir = valor * percentual_imposto_renda
    iss = valor * 0.05
    csll = valor * 0.0375
    pis = valor * 0.03
    return ir + iss + csll + pis

# Argumentos nomeados não podem ser passados antes dos argumentos posicionais, caso contrário, ocorrerá um erro de sintaxe.
# Argumento de posição(positional argument) tem que ser passado antes do argumento nomeado (keyword argument).
print(calcular_imposto(1000,percentual_imposto_renda=0.275))
```

```python
def calcular_imposto(valor, *args):
   total_imposto = 0
   print(args)
   return total_imposto

# *args permite eu passar quantos parametros de posição eu quiser, e eles serão armazenados em uma tupla.
print(calcular_imposto(1000,0.25, 0.15, 0.15, 15, 10))
```

```python
def calcular_imposto(valor, *args):
   total_imposto = 0
   for percentual in args:
       total_imposto += valor * percentual
   return total_imposto

print(calcular_imposto(1000,0.275, 0.05, 0.0375, 0.03))
```

```python
def calcular_imposto(valor, **kwargs):
   total_imposto = 0
   print(kwargs)
   return total_imposto

# **kwargs permite passar varios parâmetros nomeados (keyword arguments) para a função, e eles serão armazenados em um dicionário.
print(calcular_imposto(1000, perc_ir=0.275, perc_iss=0.05, perc_csll=0.0375, perc_pis=0.03))
```

args e kwargs são valores opcionais, ou seja, não é obrigatório passar valores para eles. Se não forem passados, eles serão uma tupla vazia e um dicionário vazio, respectivamente.
Eles também podem ser chamados de argumentos arbitrarios ou variaveis

```python
def calcular_imposto(valor, **kwargs):
   total_imposto = 0
   if 'perc_ir' in kwargs:
       total_imposto += valor * kwargs['perc_ir']
   if 'perc_iss' in kwargs:
       total_imposto += valor * kwargs['perc_iss']
   return total_imposto
print(calcular_imposto(1000, perc_ir=0.275, perc_iss=0.05, perc_csll=0.0375, perc_pis=0.03))
```

```python
def calcular_imposto(valor, **kwargs):
   total_imposto = 0
   if 'perc_ir' in kwargs:
       total_imposto += valor * kwargs['perc_ir']
   if 'perc_iss' in kwargs:
       total_imposto += valor * kwargs['perc_iss']
   if 'perc_csll' in kwargs:
       total_imposto += valor * kwargs['perc_csll']
   if 'perc_pis' in kwargs:
       total_imposto += valor * kwargs['perc_pis']
   return total_imposto
print(calcular_imposto(1000, perc_ir=0.275, perc_iss=0.05, perc_csll=0.0375, perc_pis=0.03))
```

```python

# percentual_imposto_renda é um parâmetro padrão (default argument), caso não seja passado nenhum valor para ele, o valor padrão será utilizado.
def calcular_imposto(valor, percentual_imposto_renda=0.275):
    ir = valor * percentual_imposto_renda
    iss = valor * 0.05
    csll = valor * 0.0375
    pis = valor * 0.03
    return ir + iss + csll + pis

print(calcular_imposto(1000))
```

A diferença essencial entre parâmetro e argumento resume-se ao momento e à função deles no código: o parâmetro é a variável na definição da função (o molde), enquanto o argumento é o valor real passado na chamada da função (o conteúdo).
