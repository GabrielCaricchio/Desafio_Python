### Ternary Operators (o if em 1 linha de código)

Exemplo: Quero calcular o bônus de um vendedor da nossa loja de acordo com a regra:

- Se as vendas baterem a meta de vendas (500 unidades), então o vendedor ganha 50 reais de bônus
- Caso contrário, não ganha bônus

#### if tradicional

```python
vendas = 1000

if vendas > 500:
    bonus = 50
else:
    bonus = 0
    
print(bonus)
```

#### ternary

### Outro estilo Ternary (casos de "ou")

- Nesse exemplo eu quero criar uma função que me diga se o vendedor deve ganhar bônus
- Caso a resposta seja não (caso ele não tenha batido a meta), então a função deve me dizer quanto faltou

#### if tradicional

```python
def ganhou_bonus(vendas):
    if vendas > 500:
        return True
    else:
        return f"Faltou {500 - vendas} unidades"
print(ganhou_bonus(50))
```
